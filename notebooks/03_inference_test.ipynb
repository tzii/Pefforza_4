# Connect 4 AI - Inference Test

Load a trained model and ask it what to do for specific board configurations.

```python
import sys
import os
import numpy as np
from stable_baselines3 import PPO

sys.path.append(os.path.abspath('..'))
```

### 1. Load Model

```python
model_path = "../pefforza/agent/models/notebook_model.zip" 
if os.path.exists(model_path):
    model = PPO.load(model_path)
    print("Model loaded.")
else:
    print(f"Model not found at {model_path}. Please train first.")
    model = None
```

### 2. Manual Test Cases
Create a board state where the AI *should* block or win.
0 = Empty, 1 = Self (AI), 2 = Opponent.

```python
def predict_move(board):
    if model is None: 
        return "No Model"
    action, _ = model.predict(board, deterministic=True)
    return action

# Case 1: Opponent (2) has 3 in a row horizontally. AI (1) should block.
board = np.zeros((6, 7), dtype=np.int8)
# Row 5 (bottom), columns 0, 1, 2 are Opponent
board[5, 0] = 2
board[5, 1] = 2
board[5, 2] = 2

print("Board State:")
print(board)

action = predict_move(board)
print(f"\nAI Recommends Column: {action}") # Expecting 3 (or maybe 0-2 if valid) to block?
```
